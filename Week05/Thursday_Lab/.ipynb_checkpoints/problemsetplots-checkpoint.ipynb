```python
import numpy as np
import matplotlib.pyplot as plt

data = np.array([1, 1.75, 2, 3, 3.25, 4])
n = len(data)
```

```python
x = np.sort(data)
y = np.arange(1, n + 1) / n

plt.step(x, y, where="post")
plt.xlabel("x")
plt.ylabel("ECDF")
plt.title("Empirical CDF")
plt.ylim(0, 1.1)
plt.show()
```
