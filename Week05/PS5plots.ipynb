```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt


x = [1, 1.75, 2, 3, 3.25, 4]
```

```python
sns.ecdfplot(x)
```

```python
h = 0.2

grid = np.linspace(0.5, 4.5, 1000)

density = np.zeros(len(grid))

for xi in x:
    density += ((grid >= xi - h) & (grid <= xi + h)) / (2 * h)

density /= len(x)

plt.plot(grid, density)
plt.xlabel("x")
plt.ylabel("Density")
plt.title("Uniform KDE, h = 0.2")
plt.show()
```

```python
sd = np.std(x, ddof=1)
n = len(x)

h_silverman = 1.84 * sd * n**(-0.2)

print(h_silverman)
```

```python
h = h_silverman

grid = np.linspace(0.5, 4.5, 1000)

density = np.zeros(len(grid))

for xi in x:
    density += ((grid >= xi - h) & (grid <= xi + h)) / (2 * h)

density /= len(x)

plt.plot(grid, density)
plt.xlabel("x")
plt.ylabel("Density")
plt.title("Uniform KDE, Silverman bandwidth")
plt.show()
```
